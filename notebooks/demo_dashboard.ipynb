```python
# Demo: NDVI timeseries de la parcela de Panamá
import duckdb, pandas as pd, plotly.express as px
from pathlib import Path

DB = Path.home() / ".fastclime/data/catalog.db"
df = duckdb.query("""
    SELECT date, NDVI FROM plant_ndvi_daily
    WHERE parcel_id = 'panama_demo'
    ORDER BY date
""").to_df()

fig = px.line(df, x="date", y="NDVI", title="NDVI diario • Panamá")
fig.show()
```

```python
# Comparar lámina aplicada vs. recomendada (M2 vs M3)
df = duckdb.query("""
    SELECT d.ts, d.lamina_mm as applied, p.lamina_mm as predicted
    FROM deficit_proj d
    JOIN predictions p USING(ts)
    WHERE zone_id = 'panama_zone_a'
""").to_df()

fig = px.area(df, x="ts", y=["applied","predicted"],
              labels={"value":"Lámina (mm)","variable":""},
              title="Applied vs Predicted Irrigation")
fig.show()
```
